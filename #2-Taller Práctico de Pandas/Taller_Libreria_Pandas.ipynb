#**Taller Práctico de Pandas y Operaciones Matemáticas**



```python
import pandas as pd

#--------------------PUNTO 1---------------------------------------
# Cargar el archivo Excel en un DataFrame
datos = pd.read_excel('productos_tecnologicos.xlsx')

# Mostrar las primeras filas del DataFrame
print("\n5 primeros registros del DataFrame:\n")
print(datos.head(5))

# Mostrar las últimas filas del DataFrame
print("\n10 últimos registros del DataFrame:\n")
print(datos.tail(10))

#Tipos de datos de cada columna
print("\nTipos de datos de cada columna:\n")
print(datos.dtypes)

# Tamaño del DataFrame
print("\nTamaño del DataFrame (filas, columnas):\n")
print(datos.shape)

# Descripción estadística del DataFrame
print("\nDescripción estadística de las columnas numéricas:\n")
print(datos.describe())
#-----------------------PUNTO 2-----------------------------------------------

# Nueva columna "Total_ventas"
datos["Total_Ventas"] = datos["Precio"] * datos["Cantidad"]
print("\nDataFrame con la nueva columna 'Total_Ventas':\n")
print(datos.head())

# Nueva columna "Costo_Total"
datos["Costo_Total"] = datos["Total_Ventas"] + datos["Costo_Envio"]
print("\nDataFrame con la nueva columna 'Costo_Total':\n")
print(datos.head())

#Promedio total ventas sin decimales
promedio = datos["Total_Ventas"].sum() / datos["Total_Ventas"].count()
print("\nPromedio de Total_Ventas:\n")
promedio = int(promedio)
print(promedio)

# Producto con mayor Total_Venta.

# Utilizamos idxmax() para obtener el índice del producto
# con el mayor valor en la columna "Total_Ventas"
idx = datos["Total_Ventas"].idxmax()
productoMayor = datos.loc[idx, "Producto"] # Obtenemos el nombre del producto utilizando el índice
ventasMayor = datos.loc[idx, "Total_Ventas"] # Obtenemos el valor de Total_Ventas utilizando el mismo índice
print("\nProducto con mayor valor de Total_Ventas:\n")
print(f"{productoMayor} - Total Ventas: {ventasMayor}")

#Lo anterior se repite para obtener el producto con menor Total_Ventas,
# pero utilizando idxmin() en lugar de idxmax()

# Producto con menor Total_Venta.
idx = datos["Total_Ventas"].idxmin()
productoMenor = datos.loc[idx, "Producto"]
ventasMenor = datos.loc[idx, "Total_Ventas"]
print("\nProducto con menor valor de Total_Ventas:\n")
print(f"{productoMenor} - Total Ventas: {ventasMenor}")

#----------------------PUNTO 3-------------------------------------

# Nueva columna "IVA"
datos["IVA"] = datos["Total_Ventas"] * 0.19 # Calculamos el IVA como el 19% del Total_Ventas
print("\nNueva columna 'IVA':\n")
print(datos.head())
print(".")
print(".")
print(".")
print(datos.tail())

# Nueva columna "Ganancia_Estimada"
datos["Ganancia_Estimada"] = datos["Total_Ventas"] * 0.25 # Calculamos la ganancia estimada como el 25% del Total_Ventas
print("\nNueva columna 'Ganancia_Estimada':\n")
print(datos.head())
print(".")
print(".")
print(".")
print(datos.tail())

# Nueva columna "Perdida_Estimada"
datos["Perdida_Estimada"] = datos["Total_Ventas"] * 0.0005 # Calculamos la pérdida estimada como el 0.05% del Total_Ventas
print("\nNueva columna 'Perdida_Estimada':\n")
print(datos.head())
print(".")
print(".")
print(".")
print(datos.tail())

#Porcentaje de cada producto en relación al total de ventas
calculo_porcentaje_productos = datos[["Producto", "Total_Ventas"]].copy() # Creamos un nuevo DataFrame con las columnas "Producto" y "Total_Ventas"
calculo_porcentaje_productos["Porcentaje_Ventas_(%)"] = (
    datos["Total_Ventas"] / datos["Total_Ventas"].sum()
) * 100
calculo_porcentaje_productos["Porcentaje_Ventas_(%)"] = calculo_porcentaje_productos["Porcentaje_Ventas_(%)"].round(1)
print("\nPorcentaje de cada producto en relación al total de ventas:\n")
print(calculo_porcentaje_productos.head())
print(".")
print(".")
print(".")
print(calculo_porcentaje_productos.tail())

#Productos con porcentaje de ventas mayor a 15%
productos_mayor_porcentaje_ventas = calculo_porcentaje_productos[calculo_porcentaje_productos["Porcentaje_Ventas_(%)"] > 15.00]
print("\nProductos con porcentaje de ventas mayor a 15%:\n")
if productos_mayor_porcentaje_ventas.empty:
    print("No hay productos con un porcentaje de ventas mayor a 15%.\n")
else:
    print(productos_mayor_porcentaje_ventas.head(10))

#-----------------------PUNTO 4-------------------------------------

# Calcular la media, mediana y moda de la columna Precio.
media_precio = datos["Precio"].sum() / datos["Precio"].count()
media_precio = media_precio.round()
mediana_precio = datos["Precio"].median()
mediana_precio = mediana_precio.round()
moda_precio = datos["Precio"].mode()[0]

print(f"\nMedia de Precio: {media_precio}")
print(f"Mediana de Precio: {mediana_precio}")
print(f"Moda de Precio: {moda_precio}")

# Calcular la desviación estándar de la columna Cantidad.
desviacion_cantidad = datos["Cantidad"].std()
print(f"\nDesviación estándar de Cantidad: {desviacion_cantidad}")

# Calcular el valor máximo y mínimo de Total_Venta
max_total_ventas = datos["Total_Ventas"].max()
min_total_ventas = datos["Total_Ventas"].min()
print(f"\nValor máximo de Total_Ventas: {max_total_ventas}")
print(f"Valor mínimo de Total_Ventas: {min_total_ventas}")


#-----------------------PUNTO 5-------------------------------------

#Productos de Categoría "Tecnología"
categorias = datos[datos["Categoría"] == "Tecnología"]
print("\nProductos de la categoría 'Tecnología':\n")
print(categorias.head())
print(".")
print(".")
print(".")
print(categorias.tail())

#Productos de mayor a menor segun total_ventas
productos_ordenados = datos.sort_values(by="Total_Ventas", ascending=False)
print("\nProductos ordenados de mayor a menor según Total_Ventas:\n")
print(productos_ordenados.head())
print(".")
print(".")
print(".")
print(productos_ordenados.tail())
```

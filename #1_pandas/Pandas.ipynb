# **Pandas**
es una biblioteca de programación en Python que se usa para el análisis de datos y manipulación de datos en forma tabular. Esta herramienta se ha vuelto muy popular en la comunidad de científicos de datos, analistas y desarrolladores debido a su amplia gama de funciones y su capacidad para trabajar con datos en formato de tabla de manera eficiente.

Ademas, proporciona estructuras de datos flexibles como DataFrames, que son similares a una tabla de Excel, de esta forma permite la manipulación y análisis de datos de manera sencilla.

Fue creado por Wes McKinney en 2008 por la necesidad de manejar datos de manera eficiente en Python. McKinney, un analista cuantitativo, se encontraba con dificultades al trabajar con datos en este lenguaje debido a la falta de herramientas especializadas en el manejo de datos tabulares. En respuesta a esta problematica, McKinney comenzó a desarrollar Pandas como un proyecto de código abierto mientras trabajaba en una empresa de gestión de inversiones, con el objetivo de crear una herramienta poderosa y flexible para el análisis de datos.

### **¿Para qué sirve?**

Pandas sirve para realizar todo el flujo de trabajo con datos de forma eficiente y en un solo lugar. Sus aplicaciones principales incluyen:

- Carga y lectura de datos: Puede importar datos desde una enorme variedad de formatos, como CSV, Excel, JSON, SQL, HTML y Parquet.

- Limpieza y tratamiento de datos (Data Wrangling): Es su fuerte. Permite manejar valores faltantes, eliminar duplicados, transformar variables y corregir inconsistencias.

- Análisis Exploratorio de Datos (EDA): Ofrece métodos rápidos para calcular estadísticas descriptivas (media, mediana, desviación estándar, etc.) y explorar la estructura de los datos.

- Manipulación de datos: Facilita filtrar, ordenar, agrupar, combinar y unir conjuntos de datos de manera intuitiva.

- Preparación para Machine Learning: Es la herramienta estándar para limpiar y transformar datos antes de alimentar modelos de Machine Learning con librerías como Scikit-learn.

### **Principales Funciones y Características**

Las características clave que definen a Pandas giran en torno a sus dos estructuras de datos fundamentales:

- Series: Es un array unidimensional etiquetado, similar a una columna en una hoja de cálculo. Puede contener cualquier tipo de dato.

- DataFrame: Es la estructura más importante. Es una tabla bidimensional con filas y columnas etiquetadas. Se puede pensar en un DataFrame como una colección de Series que comparten un mismo índice.

Entre sus funciones más importantes se encuentran:

**Lectura/Escritura:**

Importar/exportar datos a múltiples formatos.

*Ejemplo:* pd.read_csv(), df.to_excel()

**Inspección:**

Visualizar las primeras/últimas filas y obtener un resumen estadístico rápido.

Ejemplo: df.head(), df.tail(), df.info(), df.describe()

**Selección:**

Acceder a columnas o filas específicas por nombre o por posición.

df['columna'], df.loc[], df.iloc[]

**Filtrado**

Filtrar filas basándose en una condición booleana.

ejemplo: df[df['edad'] > 18]

**Limpieza:**

Eliminar o rellenar valores faltantes y eliminar datos duplicados.

ejemplo: df.dropna(), df.fillna(0), df.drop_duplicates()

**Operaciones:**

Realizar operaciones aritméticas entre columnas o aplicar funciones personalizadas.

ejemplo: df['col1'] + df['col2'], df.apply(funcion)

**Agrupación:**

Agrupar datos para aplicar funciones de agregación (suma, media, etc.).

ejemplo: df.groupby('categoria').mean()

### **Pandas en la Inteligencia Artificial y el Análisis de Datos**

Pandas es una pieza fundamental en el ecosistema de la IA y el análisis de datos por varias razones :

Unión con el ecosistema Python: Actúa como el puente entre la manipulación de datos (Pandas), el cálculo numérico (NumPy), la visualización (Matplotlib, Seaborn) y el machine learning (Scikit-learn, TensorFlow).

Eficiencia en el "Data Wrangling": En la mayoría de los proyectos de IA, el 80% del tiempo se dedica a limpiar y preparar los datos. Pandas automatiza y agiliza este proceso, permitiendo a los científicos de datos enfocarse en el modelo predictivo .

Trabajo con datos reales: Los datos del mundo real vienen en formatos desordenados y heterogéneos. Pandas provee todas las herramientas necesarias para manejar estas imperfecciones. Sin Pandas, trabajar con datos en Python sería tedioso y lento.

#**Ejercicios con Pandas**

##**1. Crear un DataFrame**

se crea un DataFrame a partir de un diccionario de Python. Las claves del diccionario ('Titulo', 'Genero', etc.) se convierten automáticamente en los nombres de las columnas, y cada lista es el contenido de la columna.

Codigo:


```python
import pandas as pd

# 1. Creamos un diccionario con los datos
datos_videojuegos = {
    'Titulo': ['The Legend of Zelda', 'Elden Ring', 'Hades', 'Stardew Valley', 'Cyberpunk 2077'],
    'Genero': ['Aventura', 'RPG', 'Roguelike', 'Simulación', 'RPG'],
    'Anyo': [2017, 2022, 2020, 2016, 2020],
    'Puntuacion': [97, 96, 93, 89, 86],
    'Precio': [59.99, 59.99, 24.99, 14.99, 29.99]
}

# 2. Convertimos el diccionario en un DataFrame
df_videojuegos = pd.DataFrame(datos_videojuegos)

# 3. Mostramos el DataFrame
print("--- DataFrame Creado ---")
print(df_videojuegos)
```

##**2. Operaciones entre Columnas**

se crea una nueva columna realizando una operación aritmética sencilla entre dos columnas existentes ('Precio' y 'Puntuacion'). Pandas aplica la operación elemento a elemento. Luego, se crea otra columna aplicando un factor (0.8) a toda la columna 'Precio'. Esta capacidad vectorizada es lo que hace a Pandas tan rápido y eficiente.

Codigo:

```python
# Calcular una nueva columna: Precio por punto de puntuación
# (Un precio más bajo por punto significa mejor relación calidad-precio)
df_videojuegos['Precio_por_Punto'] = df_videojuegos['Precio'] / df_videojuegos['Puntuacion']

print("\n--- DataFrame con Nueva Columna ---")
print(df_videojuegos)

# Podemos hacer operaciones más complejas, como aplicar un descuento
df_videojuegos['Precio_Oferta'] = df_videojuegos['Precio'] * 0.8 # 20% de descuento

print("\n--- Precio con Descuento ---")
print(df_videojuegos[['Titulo', 'Precio', 'Precio_Oferta']])
```

##**4.Trabajar con Datos Sintéticos y Filtrado**

**Datos Sintéticos:** Usamos pd.date_range para crear fechas y np.random.choice para seleccionar aleatoriamente entre opciones, creando un dataset realista.

**Filtrado:** Creamos una condición lógica (filtro) y la usamos dentro de [] para obtener solo las filas que la cumplen. Es una de las operaciones más poderosas de Pandas .

**Agrupación (groupby):** df.groupby('Producto')['Total_Venta'].sum() agrupa todas las filas por el nombre del producto y luego suma los valores de la columna 'Total_Venta' para cada grupo. Es equivalente a una tabla dinámica en Excel .

**apply():** Es una función muy flexible que nos permite aplicar cualquier función, incluso una función lambda (una función anónima y pequeña), a los datos de una columna o fila .

Codigo:

```python
import pandas as pd
import numpy as np

# 1. Generar datos sintéticos de ventas
np.random.seed(123) # Para que los resultados sean reproducibles
fechas = pd.date_range('2023-01-01', periods=100, freq='D')
productos = ['Laptop', 'Mouse', 'Teclado', 'Monitor']
ventas_df = pd.DataFrame({
    'Fecha': np.random.choice(fechas, size=200),
    'Producto': np.random.choice(productos, size=200),
    'Cantidad': np.random.randint(1, 15, size=200),
    'Precio_Unitario': np.random.choice([250, 15, 45, 180], size=200)
})
ventas_df['Total_Venta'] = ventas_df['Cantidad'] * ventas_df['Precio_Unitario']

print("--- Primeras 5 filas de datos sintéticos ---")
print(ventas_df.head())

# 2. Filtrado de datos: Encontrar ventas de Laptops mayores a 1000€
filtro = (ventas_df['Producto'] == 'Laptop') & (ventas_df['Total_Venta'] > 1000)
ventas_laptops_grandes = ventas_df[filtro]

print("\n--- Ventas de Laptops con total > 1000€ ---")
print(ventas_laptops_grandes)

# 3. Agrupar y analizar: Calcular el total vendido por producto
print("\n--- Total de Ingresos por Producto ---")
ingresos_por_producto = ventas_df.groupby('Producto')['Total_Venta'].sum()
print(ingresos_por_producto)

# 4. Aplicar funciones: Calcular el IVA (21%) de cada venta
ventas_df['IVA'] = ventas_df['Total_Venta'].apply(lambda x: x * 0.21)

print("\n--- Primeras filas con la columna IVA ---")
print(ventas_df[['Producto', 'Total_Venta', 'IVA']].head())
```
